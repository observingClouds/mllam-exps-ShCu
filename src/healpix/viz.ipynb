```python
import healpy as hp
import numpy as np
import matplotlib.pyplot as plt

# Set the HEALPix order
H = 4
K = 3

# Calculate the number of pixels
nside = 2**H  # nside is 2^H
npix = hp.nside2npix(nside)

# Create a HEALPix map (e.g., initialize with zeros)
healpix_map = np.zeros(npix)

# Optionally, fill the map with some data
# For example, let's fill it with random values
healpix_map = np.arange(npix)

# Plot the HEALPix map
# Mollweide projection
# hp.mollview(healpix_map, title="HEALPix Map (Mollweide)", unit="Random Values", cmap='jet')

# Cartesian projection
hp.cartview(healpix_map, title="HEALPix Map (Cartesian)", unit="Random Values", cmap='jet')

# Gnomonic projection
# hp.gnomview(healpix_map, title="HEALPix Map (Gnomonic)", unit="Random Values", cmap='jet')

# Orthographic projection
# hp.orthview(healpix_map, title="HEALPix Map (Orthographic)", unit="Random Values", cmap='jet')

# Azimuthal equidistant projection
# hp.azeqview(healpix_map, title="HEALPix Map (Azimuthal Equidistant)", unit="Random Values", cmap='jet')
plt.show()
```
